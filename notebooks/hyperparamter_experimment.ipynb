```python
from models.ann_model import ANNModel



configs = [
    ((50, 50), 0.01, 500),
    ((100,), 0.001, 200),
    ((30, 30, 30), 0.005, 1000),
]

for config in configs:
    hidden_layers, learning_rate, max_iter = config
    model = ANNModel(hidden_layer_sizes=hidden_layers, learning_rate=learning_rate, max_iter=max_iter)
    model.load_data("data/dataset.csv")
    model.preprocess_data()
    model.train()
    accuracy = model.test()
    print(f"Config: {config}, Accuracy: {accuracy:.4f}")
```
